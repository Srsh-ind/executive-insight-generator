```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    device_map="auto"
)

response = generator(
    "What are three business risks of increasing churn?",
    max_new_tokens=100
)

print(response[0]["generated_text"])
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-3B-Instruct",
    device_map="auto"
)
```

```python
prompt = """
You are a senior business analyst.

Business metrics:
Total revenue: $1,015,000
Growth: 37.5%
Top region: NA
Average churn: 5.25%

Return ONLY the following format:

INSIGHTS:
- ...
- ...
- ...

RISKS:
- ...
- ...

RECOMMENDATIONS:
- ...
- ...
- ...
"""

response = generator(
    prompt,
    max_new_tokens=200,
    do_sample=False
)

print(response[0]["generated_text"])
```
